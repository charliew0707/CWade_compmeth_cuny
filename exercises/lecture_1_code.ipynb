```python
# calculate free fall from height of tower: 


from math import sqrt

# a function that gives the height of a ball and determine time it takes to hit the ground
def time_to_fall(h,g=9.8):
    return round(sqrt((2*h)/g),2)

def main():
    height = float(input("What height is your tower: "))
    print(f"{time_to_fall(height)} seconds")

if __name__ == "__main__":
    main()
```
