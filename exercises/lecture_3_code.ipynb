```python
#Write a program that takes 3 numbers, a,b,c and prints out two solutions to quadrtic eq:
# Use to compute 0 = 0.001x^2 + 1000x + 0.001, shows two different ways to get quad formula, accuracy dependent



from math import sqrt

def quad_eq(a,b,c):
    sol_1 = (-b + sqrt(b**2 - 4*a*c)) / (2*a)
    sol_2 = (-b - sqrt(b**2 - 4*a*c)) / (2*a)
    return sol_1, sol_2

def quad_eq_2(c,d,e):
    sol_3 = 2*e / (-d + sqrt(d**2 - 4*c*e))
    sol_4 = 2*e / (-d - sqrt(d**2 - 4*c*e))
    return sol_3, sol_4

def main():
    print(f'{quad_eq(0.001,1000,0.001)}')
    print(f'{quad_eq_2(0.001,1000,0.001)}')

main()

```

```python
#Matrix Calculation to show speed, this can get up to many calculatins very fast, typical size limit is N = 1000:

# This code snippet does two operations, multiplication and addition calculation. This is done
# N times over k, N times over j and N times over i so a total of 2N3 operations.
# note about better way (np) to do matric calcs, see documentation

import numpy as np
import time

N = 200 

# Create two random matrices to multiply
A = np.random.rand(N, N)
b_matrix = np.random.rand(N, N)

C = np.zeros([N, N], float) 

start_time = time.time()

# The O(N^3) nested loop calculation
for i in range(N):
    for j in range(N):
        for k in range(N):
            C[i, j] += A[i, k] * b_matrix[k, j]

print(f"Loop method took: {time.time() - start_time:.4f} seconds")
```

```python
# QHO write a program to solve the average energy in a qho which has energy levels E_n = hf(n+1/2)

# avg energy <E> is given by 1/Z sum to inf E_n e ^ - Beta E_N, Z = sum 0 to inf e ^ -Beta E_n

# to make this doable numerically, we cannot let the sum go to inf, good chance that going to 1 million or 1 billion 
# this will still be accurate: sum do as a for loop, better way to do this without for loop

import numpy as np

hf = 1 
B = 0.01
steps = 100000
E = 0.0
Z = 0.0

for i in range(steps):
    Z += np.exp(-B*hf*(i + 1/2))

for j in range(steps):
        E += (1/Z) * hf * (j + 1/2) * np.exp(-B*hf*(j + 1/2))


print(E)
```
