```python
# Start with the code for a 3-Sphere, check answer, then make D = 10 
import numpy as np 

# Constants: 
N = 10000000
D = 10
lbound = -1
ubound = 1

def f(array): 
    # return the squared part of the x,y,z coord, check inside sphere (true or false) and then sum the boolean to get total. 
    squared = np.sum(array**2, axis=1)
    return np.sum(squared <= 1.0)



# create the array:
r_array = np.random.uniform(lbound, ubound, (N,D))



f_values = f(r_array)
V = (ubound - lbound)**D  
I = V/N * np.sum(f_values)

print(I)




```

```python
# Monte Carlo for Ideal Gas
import matplotlib.pyplot
import numpy as np
import random as rd 

T = 10.0
N = 1000
steps = 250000

# Create a 2D array to store the quantum numbers
n = np.ones([N,3],int)

# Main loop
eplot = []
E = 3*N*np.pi*np.pi/2

for k in range(steps):
 # Choose the particle and the move
    i = np.randrange(N)
    j = np.randrange(3)
    if np.random()<0.5:
        dn = 1
        dE = (2*n[i,j]+1)*np.pi*np.pi/2
    else:
        dn = -1
        dE = (-2*n[i,j]+1)*np.pi*np.pi/2
 # Decide whether to accept the move
    if n[i,j]>1 or dn==1:
        if np.random()<np.exp(-dE/T):
            n[i,j] += dn
            E += dE
            
    eplot.append(E)

# Make the graph
plt.plot(eplot); 
plt.ylabel('Energy'); 
plt.show() 
```
