```python
import numpy as np
from matplotlib.pyplot import plot, show, grid

p = np.poly1d([924,-2772,3150,-1680,420,-42,1])
z = np.linspace(0,1,500)

y = p(z)

plot(z,y)
grid()
show()

# Newton's Method:

def newton_method(x, delta=1,accuracy=1e-14,steps=1000):

    for _ in range(steps):
        delta = p(x)*(1/p.deriv()(x))
        x -= delta
    
        if abs(delta) < accuracy:
            return float(round(x,15))
            
    else:
        return None
     

    


xpoints = np.array([0.01,0.2,0.35,0.6,0.8,0.99])
roots = []
for i in xpoints:
    roots.append(newton_method(i))

print(roots)
print(np.round(p(roots),4)) 



```
