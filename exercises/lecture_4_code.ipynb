```python
# Code from Newman Example

N = 1000
a = 0 
b = 2 
h = (b-a)/N

# define the function we integrate:

def f(n):
    return n**4-2*n+1

# define the endpoints separate from the sum

s = 0.5*f(a)+0.5*f(b)

# for loop for the integral, update the sum each time, also includes endpoints already

for k in range(1,N):
    s += f(a+k*h)

# print the F = h(endpoints + sum), where s = endpoints + sum 
print(h*s)
```

```python
# Do the integral above again with np array, Will be quicker:

import numpy as np


def f(x):
    return x**4 - 2*x + 1

# function that takes endpoints and step number and integrates via trapezoidal rule, with np

def int(a,b,N):

    #define the dx or the height relation
    h = (b-a)/N
    
    #create an np array, where this gives a as start, b as endpoint, with n+1 points between (equally spaced)
    x = np.linspace(a, b, N + 1)

    # pass the entire array of spaced points into the integrand ie the function, does sequentially
    y = f(x)

    #trapezoidal formula, where y[0] = f(a) endpoint and y[-1] is f(b), sliced to end of array, np.sum then sums all non-endpoints 
    # where y[] is up to but excluding so y[0] and y[-1] are not included
    s = 0.5 * y[0] + 0.5 * y[-1] + np.sum(y[1:-1])

    return float((h * s))



int(0,2,1000)

```

```python
# now use the same function using Simpsons Method (x^2 fit to f(x):

N = 100
a = 0 
b = 2 
h = (b-a)/N

def f(n):
    return n**4 - 2*n + 1

# 1. Define the endpoints (f(a) + f(b))
s_endpoints = f(a) + f(b)
s_odds = 0 
s_evens = 0

# 2. Loop through ODD indices 
for k in range(1, N, 2):
    s_odds += f(a + k*h)

# 3. Loop through EVEN indices
for k in range(2, N, 2):
    s_evens += f(a + k*h)

# 4. Combine terms using the standard Simpson's Rule formula
integral = (h/3) * (s_endpoints + 4*s_odds + 2*s_evens)

print(integral)

```

```python
# Adaptive Integration Example: for future reference

import numpy as np

def f(x):
    return x**4 - 2*x + 1

# 1. Reuse our clean NumPy integration function from earlier
def trap_integral(a, b, N):
    h = (b - a) / N
    x = np.linspace(a, b, N + 1)
    y = f(x)
    s = 0.5 * y[0] + 0.5 * y[-1] + np.sum(y[1:-1])
    return (h * s).item()

# 2. Adaptive control function
def adaptive_trap(a, b, target_accuracy):
    N = 10                  # Start with a small number of steps
    error = 1.0             # Set an initial placeholder error greater than target
    
    # Calculate our first rough estimate
    old_integral = trap_integral(a, b, N)
    
    # Keep looping until the change between steps is smaller than our target accuracy
    while error > target_accuracy:
        N = N * 2           # Double the number of slices for the next try
        
        # Calculate the new, more precise integral
        new_integral = trap_integral(a, b, N)
        
        # Newman's error estimation rule for Trapezoidal: 
        # The true error is roughly 1/3 of the difference between the two steps
        error = (1/3) * abs(new_integral - old_integral)
        
        print(f"N = {N:4d} | Integral = {new_integral:.6f} | Estimated Error = {error:.2e}")
        
        # Update the old value to prepare for the next loop
        old_integral = new_integral
        
    return new_integral, N

# Run the adaptive integration with a target accuracy of 1e-6 (0.000001)
final_answer, final_N = adaptive_trap(0, 2, 1e-6)

print(f"\nSuccess! Final Answer: {final_answer:.6f} using {final_N} steps.")


```

```python
# Romberg Integration Example: for future reference

import numpy as np

def f(x):
    return x**4 - 2*x + 1

# Reusable trapezoidal function from before
def trap_integral(a, b, N):
    h = (b - a) / N
    x = np.linspace(a, b, N + 1)
    y = f(x)
    s = 0.5 * y[0] + 0.5 * y[-1] + np.sum(y[1:-1])
    return h * s

def romberg_integral(a, b, max_steps=5):
    # Create a blank matrix to store the triangular Romberg table rows and columns
    R = np.zeros((max_steps, max_steps))
    
    # Fill the first column (col 0) with standard Trapezoidal results by doubling N
    for i in range(max_steps):
        N = 2**i
        R[i, 0] = trap_integral(a, b, N)
        
    # Fill the rest of the table using the Richardson Extrapolation formula
    for j in range(1, max_steps):
        for i in range(j, max_steps):
            factor = 4**j
            R[i, j] = R[i, j-1] + (R[i, j-1] - R[i-1, j-1]) / (factor - 1)
            
    # Print the full matrix layout to visually see accuracy increase down and right
    print("Romberg Table:\n", R)
    
    # Return the bottom rightmost value, which holds the highest-order cancellation accuracy
    return R[-1, -1]

# Run the integration from 0 to 2
answer = romberg_integral(0, 2, max_steps=5)
print(f"\nFinal Extrapolated Answer: {answer:.6f}")


```
