```python
# Exercise 1: Show that numbers between 0 and 3, cos^2(x)+sin^2(x) = 1: 

import numpy as np

vec = np.array([0,1,2,3])

equ = np.cos(vec)**2 + np.sin(vec)**2

if  equ.all() == 1:
    print("All good")
```
