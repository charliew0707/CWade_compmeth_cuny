```python
# Dicrete Fourier Transform:
# math does not have complex, use cmath and/or np

from numpy import zeros
from cmath import exp,pi

def dft(y):
    N = len(y)
    c = zeros(N//2+1,complex)
    for k in range(N//2+1):
        for n in range(N):
            c[k] += y[n]*exp(-2j*pi*k*n/N)
    return c

```

```python
# Read file, DFT, Sunspots:

import numpy as np
from cmath import exp,pi
import matplotlib.pyplot as plt


time, reading = np.loadtxt('/Users/charleswade/Desktop/CWade_compmeth_cuny/exercises/sunspots.txt', unpack=True)

# plot the raw data from the sunspot readings
plt.plot(time, reading)
plt.show()

def dft(y):
    
    N = len(y)
    c = np.zeros(N//2+1,complex)
    for k in range(N//2+1):
        for n in range(N):
            c[k] += y[n]*exp(-2j*pi*k*n/N)
    return c

# now DFT the function:
spec = dft(reading)
mod_spec = abs(spec)**2
k_values = np.arange(len(spec))

plt.plot(k_values, mod_spec)
plt.show()


```

```python
# Using the numpy fft method: 

from numpy.fft import fft, ifft, rfft

# use sunspot data from above, pass the readings to the fft:

transformed_y = abs(rfft(reading))**2
x_k = np.arange(len(transformed_y))

plt.plot(x_k,transformed_y)
plt.show()
```
