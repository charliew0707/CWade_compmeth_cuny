````python
```python
"""
============================================================
scipy.integrate — Integration Methods Reference
============================================================

IMPORT BLOCK:
    from scipy.integrate import (
        quad, dblquad, tplquad,
        trapezoid, simpson, fixed_quad
    )

NOTE: romberg and quadrature removed in recent scipy versions.
      Use quad (replaces romberg) and fixed_quad (replaces quadrature)

------------------------------------------------------------
1. quad(func, a, b, args=(), limit=50)
------------------------------------------------------------
   Purpose : Adaptive 1D definite integral
   Args    : func     -> callable f(x) or f(x, *args)
             a, b     -> float, limits (±np.inf allowed)
             args     -> tuple of extra args for func
             limit    -> int, max subdivisions (default 50)
   Returns : (result, abserr) -> (float, float)
   Use when: general 1D integration, handles singularities
             also replaces romberg() — more accurate and robust

------------------------------------------------------------
2. dblquad(func, a, b, gfun, hfun, args=())
------------------------------------------------------------
   Purpose : Adaptive 2D definite integral ∫∫ f(y,x) dy dx
   Args    : func     -> callable f(y, x)  ← y comes first!
             a, b     -> float, outer (x) limits
             gfun     -> float or callable g(x), inner lower (y) limit
             hfun     -> float or callable h(x), inner upper (y) limit
             args     -> tuple of extra args
   Returns : (result, abserr) -> (float, float)
   Use when: 2D integrals; inner bounds can depend on x

------------------------------------------------------------
3. tplquad(func, a, b, gfun, hfun, qfun, rfun, args=())
------------------------------------------------------------
   Purpose : Adaptive 3D definite integral ∫∫∫ f(z,y,x) dz dy dx
   Args    : func     -> callable f(z, y, x) ← innermost first!
             a, b     -> float, outermost (x) limits
             gfun     -> float or callable g(x), y lower limit
             hfun     -> float or callable h(x), y upper limit
             qfun     -> float or callable q(x,y), z lower limit
             rfun     -> float or callable r(x,y), z upper limit
             args     -> tuple of extra args
   Returns : (result, abserr) -> (float, float)
   Use when: triple integrals; bounds can depend on outer vars

------------------------------------------------------------
4. trapezoid(y, x=None, dx=1.0, axis=-1)
------------------------------------------------------------
   Purpose : Trapezoidal rule on pre-sampled data
   Args    : y        -> ndarray of function values
             x        -> ndarray of sample points (or None)
             dx       -> float, uniform spacing if x not given
             axis     -> int, axis to integrate along
   Returns : result -> float or ndarray
   Note    : Use trapezoid not trapz (trapz is deprecated)
   Use when: you have data points, not a function

------------------------------------------------------------
5. simpson(y, x=None, dx=1.0, axis=-1)
------------------------------------------------------------
   Purpose : Simpson's rule on pre-sampled data
   Args    : y        -> ndarray of function values
                         (odd number of points = best accuracy)
             x        -> ndarray of sample points (or None)
             dx       -> float, uniform spacing if x not given
             axis     -> int, axis to integrate along
   Returns : result -> float or ndarray
   Note    : Use simpson not simps (simps is deprecated)
   Use when: sampled data, more accurate than trapezoid

------------------------------------------------------------
6. fixed_quad(func, a, b, args=(), n=5)
------------------------------------------------------------
   Purpose : Fixed-order Gaussian quadrature
             replaces quadrature() from older scipy
   Args    : func     -> callable f(x)
             a, b     -> float, finite limits only
             args     -> tuple of extra args
             n        -> int, number of Gauss points (default 5,
                         increase for higher accuracy)
   Returns : (result, none) -> (float, 0.0)
             note: no error estimate returned
   Use when: smooth functions on finite intervals; you want
             explicit Gaussian quadrature at a fixed order

============================================================
QUICK COMPARISON
============================================================
   Method        | Input type  | Error est. | Infinite limits
   --------------|-------------|------------|----------------
   quad          | function    | yes        | yes
   dblquad       | function    | yes        | no
   tplquad       | function    | yes        | no
   trapezoid     | data array  | no         | no
   simpson       | data array  | no         | no
   fixed_quad    | function    | no         | no
============================================================
"""
```
````

````python
import numpy as np
from scipy.integrate import (
    quad, dblquad, tplquad,
    trapezoid, simpson
)

def f(x):
    return x**4 - 2*x + 1

quad(f, 0, 2)
quad(f,0,4)

x = np.linspace(0,2,1000)
y=f(x)

trapezoid(y,x)

````
