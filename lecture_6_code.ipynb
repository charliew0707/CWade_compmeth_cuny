```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
  return 1 + 0.5 * np.tanh(2 * x)


def g(x):
  return (1 / np.cosh(2 * x)) ** 2

# central difference derivative calculator, default range [-2,2], 100 points, and optimized h
def central_difference(f, a=-2, b=2, num_points=100, h=1e-5):
    #create an array between endpoints, and space out by number of points
  x = np.linspace(a, b, num_points)
    # calc the central diff, and do it for the whole array, 
  der = (f(x + h) - f(x - h)) / (2 * h)
    # return the x value from the array, and the y value which are the der ie the array passed through the function with the cderivative formula
  return x, der

# create a plot, plot two lines, calculated derivative and the actual antiderivative
def plot_func(x1, y1, x2, y2):
  plt.figure(figsize=(8, 5))

  # Numerical derivative (dots) vs. Analytic solution (dashed line), as the separate x/y values
  plt.plot(x1, y1, "o", label="Central Difference Method", markersize=4)
  plt.plot(x2, y2, "r--", label="Analytic Result g(x)", linewidth=2)

  plt.xlabel("x")
  plt.ylabel("df/dx")
  plt.title("Central Difference Numerical Derivative vs. Analytic Solution")
  plt.grid(True, linestyle="--", alpha=0.6)
  plt.legend()

  plt.tight_layout()
  plt.show()


# Compute numerical derivative
x1, y1 = central_difference(f)

# Compute analytic derivative over the same domain
x2 = x1
y2 = g(x2)

# Generate plot
plot_func(x1, y1, x2, y2)
```
